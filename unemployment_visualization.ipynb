# Unemployment in India: Data Visualization (Part 2)
This notebook makes 7 charts from the cleaned dataset `cleaned_data.csv`.
Every insight is based only on the numbers printed by the code above it.

```python
# Import pandas to work with table data
import pandas as pd
# Import matplotlib to draw charts
import matplotlib.pyplot as plt
# Import seaborn to make nicer charts
import seaborn as sns
# Import os (built into Python) to create folders
import os

# Create a folder called plots (no error if it already exists)
os.makedirs("plots", exist_ok=True)

# Save the dataset link in a variable
url = "https://raw.githubusercontent.com/MaharudrasinhRathore/Unemployment-in-India/main/cleaned_data.csv"

# Try to load the data from the link
try:
    # Read the CSV from the link into a table called df
    df = pd.read_csv(url)
# If loading fails, use the backup plan below
except Exception:
    # Tell the user what is happening
    print("Loading failed. Please upload cleaned_data.csv")
    # Import the Colab upload tool
    from google.colab import files
    # Open the upload box so you can choose the file
    uploaded = files.upload()
    # Read the uploaded file into the table called df
    df = pd.read_csv("cleaned_data.csv")

# Show the first 5 rows of the table
print(df.head())
# Print the shape (rows, columns)
print("Shape:", df.shape)
# Print all column names
print("Columns:", list(df.columns))
```

## Chart 1: Records per state (count plot)
This count plot shows how many rows each state has in the dataset.
It helps us check that the states are covered fairly before we compare them.

```python
# Count how many rows each state has, largest first
state_counts = df["state"].value_counts()
# Set the size of the picture (tall enough for 28 states)
plt.figure(figsize=(8, 9))
# Draw a count plot with states on the y axis, sorted by count
sns.countplot(data=df, y="state", order=state_counts.index)
# Add a title to the chart
plt.title("Number of Records per State")
# Label the x axis
plt.xlabel("Number of records")
# Label the y axis
plt.ylabel("State")
# Save the chart in the plots folder
plt.savefig("plots/1_state_count.png", bbox_inches="tight")
# Show the chart on screen
plt.show()
# Print the numbers behind the chart
print(state_counts)
# Print the smallest and largest count
print("Smallest count:", state_counts.min(), "| Largest count:", state_counts.max())
```

**Insight:** 20 of the 28 states have 28 records each, which is the largest count.
Chandigarh has only 12 and Sikkim 17, so states with fewer records may be harder to compare fairly.

## Chart 2: Distribution of unemployment rate (histogram)
This histogram shows how often each unemployment rate value appears.
It helps us see the typical range and whether a few very high values stretch the right side.

```python
# Set the size of the picture
plt.figure(figsize=(8, 5))
# Draw a histogram of the unemployment rate with 20 bars
sns.histplot(data=df, x="unemployment_rate_pct", bins=20)
# Add a title to the chart
plt.title("Distribution of Unemployment Rate")
# Label the x axis
plt.xlabel("Unemployment rate (%)")
# Label the y axis
plt.ylabel("Number of records")
# Save the chart in the plots folder
plt.savefig("plots/2_unemployment_histogram.png", bbox_inches="tight")
# Show the chart on screen
plt.show()
# Print median, mean and max of the unemployment rate
print(df["unemployment_rate_pct"].describe().round(2))
# Group the rates into ranges and count the records in each range
print(pd.cut(df["unemployment_rate_pct"], bins=[0, 5, 10, 15, 20, 100]).value_counts().sort_index())
```

**Insight:** The median unemployment rate is 8.35% but the mean is 11.79%, so a few high values pull the average up.
129 records are above 20%, and the highest value is 76.74%.

## Chart 3: Unemployment rate by area type (box plot)
This box plot compares the unemployment rate in Rural and Urban areas.
The line in each box is the median, and the dots far away are the outliers.

```python
# Set the size of the picture
plt.figure(figsize=(7, 5))
# Draw a box plot of unemployment rate for each area type
sns.boxplot(data=df, x="area_type", y="unemployment_rate_pct")
# Add a title to the chart
plt.title("Unemployment Rate: Rural vs Urban")
# Label the x axis
plt.xlabel("Area type")
# Label the y axis
plt.ylabel("Unemployment rate (%)")
# Save the chart in the plots folder
plt.savefig("plots/3_area_boxplot.png", bbox_inches="tight")
# Show the chart on screen
plt.show()
# Print count, median, mean and max for each area type
print(df.groupby("area_type")["unemployment_rate_pct"].agg(["count", "median", "mean", "max"]).round(2))
```

**Insight:** The median unemployment rate is 9.97% in Urban areas and 6.76% in Rural areas.
Urban also has a higher mean (13.17% vs 10.32%) and a higher maximum (76.74% vs 74.51%).

## Chart 4: Average unemployment rate by state (bar plot)
This bar plot ranks the states by their average unemployment rate.
It helps us see which states are highest and lowest.

```python
# Calculate the average unemployment rate for each state, highest first
state_avg = df.groupby("state")["unemployment_rate_pct"].mean().sort_values(ascending=False).round(2)
# Set the size of the picture (tall enough for 28 states)
plt.figure(figsize=(8, 9))
# Draw a horizontal bar plot of the state averages
sns.barplot(x=state_avg.values, y=state_avg.index)
# Add a title to the chart
plt.title("Average Unemployment Rate by State")
# Label the x axis
plt.xlabel("Average unemployment rate (%)")
# Label the y axis
plt.ylabel("State")
# Save the chart in the plots folder
plt.savefig("plots/4_state_average_bar.png", bbox_inches="tight")
# Show the chart on screen
plt.show()
# Print the 5 highest states
print("Top 5:")
# Show the top 5 rows
print(state_avg.head(5))
# Print the 5 lowest states
print("Bottom 5:")
# Show the bottom 5 rows
print(state_avg.tail(5))
```

**Insight:** Tripura has the highest average unemployment rate (28.35%), followed by Haryana (26.28%).
Meghalaya has the lowest (4.80%), and Odisha is next lowest (5.66%).

## Chart 5: Average unemployment rate by month (bar plot)
This bar plot shows the average unemployment rate for each month in the data.
It helps us see how the rate changes over time.

```python
# Change the date column from text to real dates
df["date"] = pd.to_datetime(df["date"])
# Calculate the average unemployment rate for each date
monthly = df.groupby("date")["unemployment_rate_pct"].mean().round(2)
# Make short month labels like "May 2019"
labels = monthly.index.strftime("%b %Y")
# Set the size of the picture
plt.figure(figsize=(10, 5))
# Draw a bar plot of the monthly averages
plt.bar(labels, monthly.values)
# Turn the month labels so they do not overlap
plt.xticks(rotation=45)
# Add a title to the chart
plt.title("Average Unemployment Rate by Month")
# Label the x axis
plt.xlabel("Month")
# Label the y axis
plt.ylabel("Average unemployment rate (%)")
# Save the chart in the plots folder
plt.savefig("plots/5_monthly_average_bar.png", bbox_inches="tight")
# Show the chart on screen
plt.show()
# Print the monthly averages
print(monthly)
```

**Insight:** The monthly average stays between 8.87% and 10.70% from May 2019 to March 2020.
It jumps to 23.64% in April 2020 and 24.88% in May 2020, then falls to 11.90% in June 2020.

## Chart 6: Labour participation vs unemployment rate (scatter plot)
This scatter plot puts each record as a dot, with participation on the x axis and unemployment on the y axis.
It helps us see whether the two values move together.

```python
# Set the size of the picture
plt.figure(figsize=(8, 6))
# Draw a scatter plot, with colour showing Rural or Urban
sns.scatterplot(data=df, x="labour_participation_rate_pct", y="unemployment_rate_pct", hue="area_type")
# Add a title to the chart
plt.title("Labour Participation Rate vs Unemployment Rate")
# Label the x axis
plt.xlabel("Labour participation rate (%)")
# Label the y axis
plt.ylabel("Unemployment rate (%)")
# Save the chart in the plots folder
plt.savefig("plots/6_scatter_participation_unemployment.png", bbox_inches="tight")
# Show the chart on screen
plt.show()
# Print the correlation between the two columns
print("Correlation:", round(df["labour_participation_rate_pct"].corr(df["unemployment_rate_pct"]), 2))
# Print the average of both columns for each area type
print(df.groupby("area_type")[["labour_participation_rate_pct", "unemployment_rate_pct"]].mean().round(2))
```

**Insight:** The correlation between participation and unemployment is 0.0, so there is no straight-line pattern.
On average, Rural has higher participation (44.46%) but lower unemployment (10.32%) than Urban (40.90% and 13.17%).

## Chart 7: Correlation heatmap
This heatmap shows how strongly the three number columns move together.
Values near 1 or -1 mean a strong link, and values near 0 mean almost no straight-line link.

```python
# Pick the three number columns
num_cols = ["unemployment_rate_pct", "estimated_employed", "labour_participation_rate_pct"]
# Calculate the correlation between them
corr = df[num_cols].corr().round(2)
# Set the size of the picture
plt.figure(figsize=(7, 5))
# Draw the heatmap and write the numbers inside the squares
sns.heatmap(corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
# Add a title to the chart
plt.title("Correlation Heatmap")
# Save the chart in the plots folder
plt.savefig("plots/7_correlation_heatmap.png", bbox_inches="tight")
# Show the chart on screen
plt.show()
# Print the correlation table
print(corr)
```

**Insight:** Unemployment rate and estimated employed have a correlation of -0.22, the strongest pair in the table.
Participation has about 0.0 with unemployment and 0.01 with employed, so those links are almost zero.
